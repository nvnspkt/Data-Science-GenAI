# Calculator
This notebook takes two numbers from the user and performs addition, subtraction, multiplication, division, floor division, modulus, and exponentiation.

## Step 1: Take Input from the User for a and b

```python
a = float(input("Enter the first number (a): "))
b = float(input("Enter the second number (b): "))
print(f"Numbers entered: a = {a}, b = {b}")
```

## Step 2: Addition

```python
print(f"Addition: {a} + {b} = {a + b}")
```

## Step 3: Subtraction

```python
print(f"Subtraction: {a} - {b} = {a - b}")
```

## Step 4: Multiplication

```python
print(f"Multiplication: {a} * {b} = {a * b}")
```

## Step 5: Division

```python
if b == 0:
    print("Cannot divide by zero.")
else:
    print(f"Division: {a} / {b} = {a / b}")
```

## Step 6: Floor Division

```python
if b == 0:
    print("Cannot divide by zero.")
else:
    print(f"Floor Division: {a} // {b} = {a // b}")
```

## Step 7: Modulus

```python
if b == 0:
    print("Cannot find modulus with zero.")
else:
    print(f"Modulus: {a} % {b} = {a % b}")
```

## Step 8: Exponentiation

```python
print(f"Exponentiation: {a} ** {b} = {a ** b}")
```
